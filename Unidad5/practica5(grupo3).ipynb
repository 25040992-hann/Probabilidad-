```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df3=pd.read_csv("grupo3.csv")
df3.head()
```

```python
print(df3.describe())
```

```python
plt.hist(df3['Calificacion'], bins=10, color="skyblue")
plt.title("Histograma de Calificaciones de Grupo 3")
plt.xlabel("Calificación")
plt.show()
```

```python
Horas = df3[["HorasEstudio"]]
Calif = df3["Calificacion"]

model = LinearRegression()
model.fit(Horas, Calif)

print("Coeficiente de la Pendiente:", model.coef_[0])
print("Intercepción de la Recta:", model.intercept_)
```

```python
correlacion = df3["HorasEstudio"].corr(df3["Calificacion"])
print("Coeficiente de correlación de Pearson:", correlacion)
```

```python
plt.figure(figsize=(6, 4))

Horas = [3.79, 6.08, 7.17, 7.54, 8.18]
Calif = [15, 26.1, 29, 31.9, 34.8]

sns.set(style="whitegrid")
sns.regplot(x=Horas, y=Calif, scatter=True, color='blue')
plt.title('Relación de Horas de estudio con Calificaciones')
plt.xlabel('Horas de Estudio')
plt.ylabel('Calificacion')
plt.grid(True)
plt.show()
```

###1. Interpretación del coeficiente de regresión
-> La pendiente es 5.285670950508442.

-> Esto significa que por cada hora adicional de estudio, la calificación final aumenta bastante. La relación es positiva y altamente marcada.

###2. Interpretación del coeficiente de correlación
-> El coeficiente de correlación de Pearson es 0.9422151342073145.

-> Indica una relación positiva fuerte entre las horas de estudio y las calificaciones, entonces a mayor tiempo de estudio, mayor es la calificación obtenida.

###3. Significancia estadística del modelo
-> Es estadísticamente significativo, ya que la correlación es muy alta y la pendiente positiva, lo que refleja una relación clara entre las variables.

###4. Confiabilidad del modelo para predecir
-> Se puede considerar confiable para predecir las calificaciones basadas en el tiempo de estudio.

-> En este grupo, el tiempo de estudio explica gran parte de la variación en las calificaciones y puede usarse como predictor sólido.

###Conclusión
-> A mayor tiempo de estudio, mayor es la calificación, y la relación es muy fuerte y positiva.

-> El tiempo de estudio puede explicar de manera efectiva las calificaciones obtenidas y el modelo es se puede considerar confiable para predecir el rendimiento académico.
