```python
df1=pd.read_csv("grupo1.csv")
df1.head()
```

```python
print(df1.describe())
```

```python
plt.hist(df1['Calificacion'], bins=10, color="skyblue")
plt.title("Histograma de Calificaciones de Grupo 1")
plt.xlabel("Calificación Final")
plt.show()
```

```python
Horas = df1[["HorasEstudio"]]
Calif = df1["Calificacion"]

model = LinearRegression()
model.fit(Horas, Calif)

print("Coeficiente de la Pendiente:", model.coef_[0])
print("Intercepción de la Recta:", model.intercept_)
```

```python
correlacion = df1["HorasEstudio"].corr(df1["Calificacion"])
print("Coeficiente de correlación de Pearson:", correlacion)
```
