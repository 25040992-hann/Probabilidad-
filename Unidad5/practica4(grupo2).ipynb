##Grupo 2

```python
df2=pd.read_csv("grupo2.csv")
df2.head()
```

```python
print(df2.describe())
```

```python
plt.hist(df2['Calificacion'], bins=10, color="skyblue")
plt.title("Histograma de Calificaciones de Grupo 2")
plt.xlabel("Calificación")
plt.show()
```

```python
Horas = df2[["HorasEstudio"]]
Calif = df2["Calificacion"]

model = LinearRegression()
model.fit(Horas, Calif)

print("Coeficiente de la Pendiente:", model.coef_[0])
print("Intercepción de la Recta:", model.intercept_)
```

```python
correlacion = df2["HorasEstudio"].corr(df2["Calificacion"])
print("Coeficiente de correlación de Pearson:", correlacion)
```

```python
plt.figure(figsize=(6, 4))

Horas = [3.79, 6.08, 7.17, 7.54, 8.18]
Calif = [40, 26.1, 29, 31.9, 34.8]

sns.set(style="whitegrid")
sns.regplot(x=Horas, y=Calif, scatter=True, color='blue')
plt.title('Relación de Horas de estudio con Calificaciones')
plt.xlabel('Horas de Estudio')
plt.ylabel('Calificacion')
plt.grid(True)
plt.show()
```

###1. Interpretación del coeficiente de regresión
-> El coeficiente de la pendiente es 4.222718653627889.

-> Esto significa que por cada hora adicional de estudio, la calificación aumenta.

-> La relación se ve positiva y fuerte.

###2. Interpretación del coeficiente de correlación
-> El coeficiente de correlación es 0.8619890426505082.

-> Indica una relación positiva entre el tiempo de estudio y las calificaciones.
Quiere decir que a mayor tiempo de estudio, mayor es la calificación.

###3. Significancia estadística del modelo
-> Es estadísticamente significativo, ya que la correlación es alta y la pendiente positiva refleja una relación clara entre las variables.

###4. Confiabilidad del modelo para predecir
-> Es confiable para predecir las calificaciones basadas en el tiempo de estudio.

-> El tiempo de estudio sí explica la variación en las calificaciones y puede usarse como predictor válido.

##Conclusión
-> El modelo de regresión lineal muestra una pendiente positiva y un coeficiente de correlación de 0.8619890426505082.

-> Quiere decir que el tiempo de estudio influye en gran parte en las calificaciones y el modelo es confiable para hacer predicciones.
