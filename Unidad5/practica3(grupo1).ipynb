##Grupo 1

```python
df1=pd.read_csv("grupo1.csv")
df1.head()
```

```python
print(df1.describe())
```

```python
plt.hist(df1['Calificacion'], bins=10, color="skyblue")
plt.title("Histograma de Calificaciones de Grupo 1")
plt.xlabel("Calificación")
plt.show()
```

```python
Horas = df1[["HorasEstudio"]]
Calif = df1["Calificacion"]

model = LinearRegression()
model.fit(Horas, Calif)

print("Coeficiente de la Pendiente:", model.coef_[0])
print("Intercepción de la Recta:", model.intercept_)
```

```python
correlacion = df1["HorasEstudio"].corr(df1["Calificacion"])
print("Coeficiente de correlación de Pearson:", correlacion)
```

```python
plt.figure(figsize=(6, 4))

Horas = [20, 15.60, 17.91, 20, 20]
Calif = [75.94, 87.48, 79.40, 73.37, 83.75]

sns.set(style="whitegrid")
sns.regplot(x=Horas, y=Calif, scatter=True, color='blue')
plt.title('Relación de Horas de estudio con Calificaciones finales')
plt.xlabel('Horas de Estudio')
plt.ylabel('Calificacion')
plt.grid(True)
plt.show()
```

###1. Interpretación del coeficiente de regresión
-> El coeficiente de la pendiente obtenido es -0.10608151145860807.

-> Esto significa que por cada hora adicional de estudio, la calificación final refleja una ligera negativa entre el tiempo de estudio y las calificaciones, lo que quiere decir que la calificación disminuye.

###2. Interpretación del coeficiente de correlación
-> El coeficiente de correlación de Pearson es  -0.05713334446646929.

-> Este valor está muy cercano a 0, lo que indica una relación casi nula entre las horas de estudio y las calificaciones.

-> La dirección es negativa, pero la fuerza es muy débil, por lo que no se puede afirmar que exista una relación lineal significativa.

###3. Significancia estadística del modelo
-> Dado que la correlación es casi cero y la pendiente es muy pequeña, el modelo de regresión no resulta estadísticamente significativo.

-> Esto puede significar que el tiempo de estudio, en estos datos, no explica de manera confiable las variaciones en las calificaciones.

###4. Confiabilidad del modelo para predecir
-> El modelo es poco confiable.

-> La pendiente negativa contradice las expectativas (más estudio = mejor calificacion).

-> La correlación débil muestra que las calificaciones pueden dependen de otros factores.


###Conclusión
El modelo de regresión lineal aplicado muestra que no existe una relación significativa entre el tiempo de estudio y las calificaciones finales. La pendiente es negativa y la correlación muy débil, por lo que se podria considerar que el modelo no es confiable para predecir el rendimiento académico.
