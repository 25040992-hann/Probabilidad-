```python
df2=pd.read_csv("grupo2.csv")
df2.head()
```

```python
print(df2.describe())
```

```python
plt.hist(df2['Calificacion'], bins=10, color="skyblue")
plt.title("Histograma de Calificaciones de Grupo 2")
plt.xlabel("Calificación Final")
plt.show()
```

```python
Horas = df2[["HorasEstudio"]]
Calif = df2["Calificacion"]

model = LinearRegression()
model.fit(Horas, Calif)

print("Coeficiente de la Pendiente:", model.coef_[0])
print("Intercepción de la Recta:", model.intercept_)
```

```python
correlacion = df2["HorasEstudio"].corr(df2["Calificacion"])
print("Coeficiente de correlación de Pearson:", correlacion)
```
